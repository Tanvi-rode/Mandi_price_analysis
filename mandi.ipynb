```python
import numpy as np
import pandas as pd
df=pd.read_csv("Raw_Mandi.csv")
df.head()
```

```python
print("columns:",df.columns.tolist())
print("\nMissing Values:\n",df.isnull().sum())
print("\nSample dates:\n",df["Arrival_Date"].unique()[:5])
```

```python
df.columns=(df.columns.str.replace('_x0020_','_').str.strip().str.replace(" ","_"))
df["Arrival_Date"]=pd.to_datetime(df["Arrival_Date"],dayfirst=True,errors="coerce")
print(f"Available Date Range: {df['Arrival_Date'].min()} to "f" {df['Arrival_Date'].max()}")
df_filtered=df[(df["Arrival_Date"]>="2026-09-01")&(df["Arrival_Date"]<="2026-11-30")].copy()
commodities=["Onion","Tomato","Potato","Rice"]
df_filtered=df_filtered[
    (df_filtered["State"].str.strip()=="Maharashtra")
    &(df_filtered["Commodity"].str.strip().isin(commodities))
    ]
print(f"Rows after filtering:{len(df_filtered)}")
```

```python
text_cols=["District","Market","Commodity","Variety"]
for col in text_cols:
 df_filtered[col]=df_filtered[col].astype(str).str.strip().str.title()
    
df_filtered = df_filtered[
    (df_filtered["Modal_Price"] > 0)
    &(df_filtered["Min_Price"] > 0)
    &(df_filtered["Max_Price"] > 0)
].drop_duplicates() 

df_filtered["Price_Spread"]=(
    df_filtered["Max_Price"]-df_filtered["Min_Price"]
)
df_filtered["Price_Per_Kg"]=(df_filtered["Modal_Price"]/100).round(2)
df_filtered.head()
```

```python
df_filtered.to_csv("clean_mandi_maharashtra.csv",index=False)
print("saved'clean_mandi_maharashtra.csv' successfully!")
```
